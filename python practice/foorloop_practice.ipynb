```python
n = 10

for i in range(1,n+1):
    print("* "*i)
```

```python
for i in range(1,11):
    for j in range(1,i+1):
        print(j,end=" ")
    print()
```

```python
n = 10

for i in range(1,n+1):
    print(" "*(n-i) + "*"*i)
```

```python
n = 10

for i in range(1,n+1):
    print(" "*(n-i) + "*"*(2*i-1))
```
